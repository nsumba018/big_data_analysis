```python
with open("supermarket_sales - Sheet1.csv","r") as f:
    # Read the column names
    headers = f.readline().strip().split(",")

    # store the values as dictionaries
    transactions = []

    for line in f:
        values = line.strip().split(",")
        records = dict(zip(headers, values))
        transactions.append(records)
# print the first 3 records
for record in transactions[:3]:
    print(record)

# print the sum of all the transactions
print("Print all the transactions: ", len(transactions))

#  print all columns names
print(headers)
    
```

QUESTION 10:

Add the following four new columns to the DataFrame. Print the first 3 rows after all columns
have been added.
a) revenue : calculated as Unit price multiplied by Quantity.
    
b) is_member : True if the Customer type is Member, False otherwise.
    
c) high_value : True if the Total is greater than 500, False otherwise.
    
d) rating_label : the value Excellent if the Rating is 8 or above, Good if it is 6 or above, and
Poor otherwise.
After adding all four columns, print how many transactions are marked as high-value and show
the count of each rating label.

```python
import numpy as np

# (a)
sales["revenue"] = sales["Unit price"] * sales["Quantity"]

# (b)
sales["is_member"] = sales["Customer type"] == "Member"

# (c)
sales["high_value"] = sales["Total"] > 500

# (d)
sales["rating_label"] = np.where(
    sales["Rating"] >= 8,
    "Excellent",
    np.where(
        sales["Rating"] >= 6,
        "Good",
        "Poor"
    )
)

# Print first 3 rows
print(sales.head(3))

# Number of high value transactions
print("High Value Transactions:", sales["high_value"].sum())

# Count each rating label
print(sales["rating_label"].value_counts())
```

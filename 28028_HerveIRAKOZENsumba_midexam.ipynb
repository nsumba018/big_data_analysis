### QUESTION 1

```python
# a)
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv")
df.to_csv("titanic.csv", index=False)
print("DataSet Saved Successfully")

# creating the dictionary
import csv

records = []
with open("titanic.csv", "r") as f:
    reader = csv.DictReader(f)

    for row in reader:
        records.append(row)
        
# b)
print("\n First three records")
for record in records[:3]:
    print(record)

print("\nTotal Number of Passengers:", len(records))

# c)
sex_set = set()

for record in records:
    sex_set.add(record["Sex"])

print("\nUnique Values in Sex Column:")
print(sex_set)

#d 
class_count = {}

for record in records:
    pclass = record["Pclass"]

    if pclass in class_count:
        class_count[pclass] += 1
    else:
        class_count[pclass] = 1

print("\nPassengers in Each Class:")
print(class_count)

# e)
survivors = []

for record in records:
    if record["Survived"] == "1":
        survivors.append(record)

with open("survivors.csv", "w", newline="") as file:
    writer = csv.DictWriter(file, fieldnames=records[0].keys())

    writer.writeheader()
    writer.writerows(survivors)

print("\nNumber of survivor records written:", len(survivors))
```

### QUESTION 2

```python
import pandas as pd 
import numpy as np

#load the dataset
df = pd.read_csv("titanic.csv")

# Remove rows with missing Age
df = df.dropna(subset=["Age"])

# Convert columns to NumPy arrays
age_arr = df["Age"].to_numpy()
fare_arr = df["Fare"].to_numpy()

# a)
print("Age Array")
print("Shape:", age_arr.shape)
print("Data Type:", age_arr.dtype)

print("\nFare Array")
print("Shape:", fare_arr.shape)
print("Data Type:", fare_arr.dtype)

# b) 
print("\nAge Statistics")
print(f"Mean: {np.mean(age_arr):.2f}")
print(f"Median: {np.median(age_arr):.2f}")
print(f"Standard Deviation: {np.std(age_arr):.2f}")

print("\nFare Statistics")
print(f"Mean: {np.mean(fare_arr):.2f}")
print(f"Median: {np.median(fare_arr):.2f}")
print(f"Standard Deviation: {np.std(fare_arr):.2f}")

# c)
young_passengers = age_arr[age_arr < 18]
print("\nPassengers younger than 18 years old: ", len(young_passengers))
```

```python
import pandas as pd

# Read the CSV file
data = pd.read_csv("titanic.csv")

# Store it as a DataFrame
df = pd.DataFrame(data)

# Display
df



# a)
print("Shape of the DataFrame:")
print(df.shape)

print("\nData Types of Each Column:")
print(df.dtypes)

print("\nFirst 5 Rows:")
display(pd.DataFrame(df.head()))

# b)
selected_df = pd.DataFrame(df[["Name", "Sex", "Age", "Pclass", "Survived"]])
print("Shape of Selected DataFrame:")
print(selected_df.shape)
display(selected_df)


# c)
highest_fare = df["Fare"].max()
highest_passenger = pd.DataFrame(df[df["Fare"] == highest_fare])
display(highest_passenger[["Name", "Pclass", "Fare"]])


# d)
first_class = pd.DataFrame(df[df["Pclass"] == 1])
print("First Class Passengers")
print("Shape:", first_class.shape)
display(first_class.head(3))

# Female Passengers Who Survived
female_survived = pd.DataFrame(
    df[(df["Sex"] == "female") & (df["Survived"] == 1)]
)
print("Female Passengers Who Survived")
print("Shape:", female_survived.shape)
display(female_survived.head(3))

# Passengers Older Than 60
older_than_60 = pd.DataFrame(df[df["Age"] > 60])

print("Passengers Older Than 60")
print("Shape:", older_than_60.shape)

display(older_than_60.head(3))

# e)
df["Family_Size"] = df["SibSp"] + df["Parch"] + 1
df["Is_Alone"] = df["Family_Size"] == 1
display(pd.DataFrame(df[["Family_Size", "Is_Alone"]].head()))

# f)
missing_values = df.isnull().sum()
missing_percentage = round((missing_values / len(df)) * 100, 2)
missing_df = pd.DataFrame({
    "Missing Values": missing_values,
    "Percentage (%)": missing_percentage
})
display(missing_df)

# g)
median_age = df["Age"].median()
df["Age"] = df["Age"].fillna(median_age)
print("Median Age Used:", median_age)
display(pd.DataFrame(df.head()))

#h)
print("Embarked Value Counts Before Filling")
print(df["Embarked"].value_counts(dropna=False))
df["Embarked"] = df["Embarked"].fillna("S")
print("\nEmbarked Value Counts After Filling")
print(df["Embarked"].value_counts(dropna=False))
display(pd.DataFrame(df.head()))

# i)
print("Columns before Removing cabins")
print(df.columns)
df = df.drop(columns=["Cabin"])
print("\nColumns after removing cabins")
print(df.columns)
display(pd.DataFrame(df.head()))
```

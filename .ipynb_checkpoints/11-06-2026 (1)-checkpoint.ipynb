Create two lists with 5 student names and
another list which is holding their respective grades
use ternary operations to decide if a student has passed
or failed. If grades is >10, the category will be Pass
else, the category will be fail

```python
names=["Elie","Eric","Lys","Alexis","David"]
scores=[45,56,75,48,85]
for name,score in zip(names,scores):
    if score>=50:
        grade="Pass"
    else:
        grade="Fail"
    print(f"Name:{name} Score:{score} Category:{grade}")
```

Account management:
We want to develop a short while loop
program that will allow a client
to check balance, deposit some money,
withdraw the money or exit the program
Below are the choices:
1. Check balance
2. Deposit: The deposit should not be negative or 0
3. Withdraw: Your balance should
   not be negative or zero
4. Exit: Stop the program
Note: The program should keep showing the options
unless a client decides to stop the program

```python
balance=5000
while True:
    choice=int(input("Enter your choice\n"))
    print("1. Check Balance")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Exit")
    if choice==1:
        print(f"The current balance is: {balance}")
    elif choice==2:
        amount=float(input("Enter the amount"))
        if amount>0:
            print(f"THe old balance is:{balance}")
            balance+=amount
            print(f"You have deposited: {amount},The new balance is:{balance}")
        else:
            print("Amount should not be zero or negative")
    elif choice==3:
        amount=float(input("Enter the amount"))
        if amount<balance:
            print(f"The old balance is: {balance}")
            balance-=amount
            print(f"The new balance is: {balance}")
        else:
            print("Amount should not be greater than or equal to the balance")
    elif choice==4:
        break
    else:
        print("Invalid choice")
        
        
        
    
```

```python
# Function with a parameter
def grade_category(marks):
    if marks>=16 and marks<=20:
        return "Grand Distinction"
    elif marks>=14 and marks<16:
        return "Distinction"
    elif marks>=12 and marks<14:
        return "Satisfaction"
    elif marks>=10 and marks<12:
        return "probation"
    elif marks>=0 and marks<10:
        return "Fail"
    else:
        return "INVALID"
#Calling The function
scores = float(input("Enter your marks"))
grade_category(scores)
```

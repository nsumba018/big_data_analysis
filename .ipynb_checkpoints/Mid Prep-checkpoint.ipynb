QUESTION 1

```python
with open("supermarket_sales - Sheet1.csv","r") as f:
    # Read the column names
    headers = f.readline().strip().split(",")

    # store the values as dictionaries
    transactions = []

    for line in f:
        values = line.strip().split(",")
        records = dict(zip(headers, values))
        transactions.append(records)
# print the first 3 records
for record in transactions[:3]:
    print(record)

# print the sum of all the transactions
print("Print all the transactions: ", len(transactions))

#  print all columns names
print(headers)
    
```

QUESTION 4: 

Load the dataset using Pandas. Extract the Unit price, Quantity, Total, and Rating columns as
NumPy arrays. Name them unit_price_arr, quantity_arr, total_arr, and rating_arr respectively.
For each array, print its shape, data type, minimum value, maximum value, and first 5 elements.

```python
import numpy as np
import pandas as pd
sales = pd.read_csv("supermarket_sales - Sheet1.csv")
unit_price = np.array(sales["Unit price"])
quantity_arr = np.array(sales["Quantity"])
total_arr = np.array(sales["Total"])
rating_arr = np.array(sales["Rating"])
# unit_price
print(unit_price.shape)
print(type(unit_price))
print(unit_price.min())
print(unit_price.max())
print(unit_price[:5])
# quantity_price
print(quantity_arr.shape)
print(type(quantity_arr))
print(quantity_arr.min())
print(quantity_arr.max())
print(quantity_arr[:5])
# total_arr
print(total_arr.shape)
print(type(total_arr))
print(total_arr.min())
print(total_arr.max())
print(total_arr[:5])
# rating_arr
print(rating_arr.shape)
print(type(rating_arr))
print(rating_arr.min())
print(rating_arr.max())
print(rating_arr[:5])
```

QUESITON 5

Using NumPy functions only, compute the following statistics for total_arr and for rating_arr.
Print all results neatly formatted to 2 decimal places.
a) Mean, median, and standard deviation.
b) Minimum and maximum values.
c) The value at the 25th percentile and the value at the 75th percentile.

```python
total_arr = np.array(sales["Total"])
rating_arr = np.array(sales["Rating"])

import numpy as np

# Statistics for Total
print("TOTAL STATISTICS")
print("Mean:", format(np.mean(total_arr), ".2f"))
print("Median:", format(np.median(total_arr), ".2f"))
print("Standard Deviation:", format(np.std(total_arr), ".2f"))
print("Minimum:", format(np.min(total_arr), ".2f"))
print("Maximum:", format(np.max(total_arr), ".2f"))
print("25th Percentile:", format(np.percentile(total_arr, 25), ".2f"))
print("75th Percentile:", format(np.percentile(total_arr, 75), ".2f"))

print()

# Statistics for Rating
print("RATING STATISTICS")
print("Mean:", format(np.mean(rating_arr), ".2f"))
print("Median:", format(np.median(rating_arr), ".2f"))
print("Standard Deviation:", format(np.std(rating_arr), ".2f"))
print("Minimum:", format(np.min(rating_arr), ".2f"))
print("Maximum:", format(np.max(rating_arr), ".2f"))
print("25th Percentile:", format(np.percentile(rating_arr, 25), ".2f"))
print("75th Percentile:", format(np.percentile(rating_arr, 75), ".2f"))
```

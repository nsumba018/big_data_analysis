Account Management


```python
balance =0
while True:
    choice = int(input("Enter your choice\n"))
    print("1. Check Balance")
    print("2. Deposit")
    print("3. WithDraw")
    print("4. Exit")
    if choice == 1:
        print(f"The current amount is:  {balance}")
    elif choice == 2:
        amount = float(input("Enter the amount: "))
        if amount > 0:
            print(f"The old balance is : {balance}")
            balance += amount
            print(f"You have deposited: {amount}, The new balance is: {balance}")
        else:
            print("Amount should not be under 0")
    elif choice == 3:
        amount = float(input("Enter the amount to withdraw"))
        if amount <balance:
            print(f"The old balance is: {balance}")
            balance -= amount
            print(f"The new balance is: {balance}")
        else:
            print("Amount should not be greater that the balance")
    elif choice == 4:
        print("Goodbye")
        break
    else:
        print("Invalid choice")
                       
            
```
